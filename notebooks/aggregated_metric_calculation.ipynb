# Aggregated metric calculation

Summary of the performance of the baseline and adapted metamodels on the total CSO volumes, from the outputs of the Snakemake workflow (`snakemake --cores <N>`):

- Accuracy metrics (MAE, MAPE and NSE) per intervention level (Table 5.1).
- Ranking preservation (RP) per intervention level (Table 5.2).
- Variability of the SWMM CSO volumes within and between intervention levels.

```python
from pathlib import Path

import pandas as pd

SAVED_OBJECTS = Path("../saved_objects")
METAMODEL_NAMES = ["baseline_metamodel", "adapted_metamodel"]
```

## Error metrics and ranking preservation

```python
for name in METAMODEL_NAMES:
    metrics = pd.read_csv(SAVED_OBJECTS / "saved_metrics" / f"metrics_total_volumes_{name}.csv")
    rp = pd.read_csv(SAVED_OBJECTS / "saved_metrics" / f"ranking_preservation_{name}.csv")
    print(name)
    display(rp.merge(metrics, on="lvl", how="left"))
```

## Variability of the SWMM CSO volumes

```python
volumes = pd.read_csv(SAVED_OBJECTS / "saved_objectives" / f"aggregated_volumes_{METAMODEL_NAMES[0]}.csv")
volumes_per_level = (
    volumes.groupby("lvl")
    .agg(
        total_volume_swmm_mean=("total_volume_swmm", "mean"),
        total_volume_swmm_std=("total_volume_swmm", "std"),
    )
    .reset_index()
)
display(volumes_per_level)

print("Standard deviation interlevels:", volumes_per_level.total_volume_swmm_mean.std())
print("Standard deviation intralevels:", volumes_per_level.total_volume_swmm_std.mean())
```
